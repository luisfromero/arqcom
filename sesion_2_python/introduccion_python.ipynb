# Introducción a Python

Python es un lenguaje de programación de alto nivel, interpretado, interactivo y orientado a objetos. Es conocido por su sintaxis clara y legible.

## 1. Variables y Tipos de Datos

En Python, no necesitas declarar explícitamente el tipo de una variable. El tipo se infiere del valor que se le asigna.

```python
# Enteros (int)
numero_entero = 10
print(f"Esto es un entero: {numero_entero}")

# Flotantes (float)
numero_flotante = 3.14
print(f"Esto es un flotante: {numero_flotante}")

# Cadenas de texto (str)
saludo = "Hola, mundo!"
print(f"Esto es una cadena: {saludo}")

# Booleanos (bool)
es_verdadero = True
print(f"Esto es un booleano: {es_verdadero}")
```

## 2. Estructuras de Control

Permiten controlar el flujo de ejecución del programa.

### Condicionales (if-elif-else)

```python
edad = 18

if edad < 18:
    print("Eres menor de edad.")
elif edad == 18:
    print("¡Justo tienes 18 años!")
else:
    print("Eres mayor de edad.")
```

### Bucles (for y while)

```python
# Bucle for: itera sobre una secuencia (como una lista)
print("Bucle for:")
for i in range(5):  # Itera de 0 a 4
    print(i)

# Bucle while: se ejecuta mientras una condición sea verdadera
print("\nBucle while:")
contador = 0
while contador < 5:
    print(contador)
    contador += 1
```

## 3. Funciones

Las funciones son bloques de código reutilizables que realizan una tarea específica.

```python
def saludar(nombre):
    """Esta función saluda a la persona pasada como argumento."""
    return f"¡Hola, {nombre}!"

# Llamamos a la función
mensaje = saludar("Ana")
print(mensaje)
```

## 4. Clases y Objetos (Introducción)

Python es un lenguaje orientado a objetos. Esto significa que permite agrupar datos (atributos) y comportamientos (métodos) en entidades llamadas **objetos**.

Una **clase** es como un plano o una plantilla para crear objetos. Define los atributos y métodos que tendrán los objetos de esa clase.

```python
# Definimos una clase llamada 'Perro'
class Perro:
    # El método __init__ es el constructor. Se llama al crear un nuevo objeto.
    def __init__(self, nombre, raza):
        # Atributos: datos asociados al objeto
        self.nombre = nombre
        self.raza = raza
        print(f"¡Guau! Se ha creado un perro llamado {self.nombre}.")

    # Método: una función que pertenece a la clase
    def ladrar(self):
        return f"{self.nombre} dice: ¡Guau, guau!"

# Creamos dos objetos (instancias) de la clase Perro
mi_perro = Perro("Fido", "Golden Retriever")
otro_perro = Perro("Luna", "Pastor Alemán")

# Accedemos a los atributos de los objetos
print(f"{mi_perro.nombre} es de raza {mi_perro.raza}.")

# Llamamos a los métodos de los objetos
print(otro_perro.ladrar())
```

En el ejemplo anterior:
- **Clase**: `Perro` es la plantilla.
- **Objeto**: `mi_perro` y `otro_perro` son instancias (objetos) creados a partir de la clase `Perro`.
- **Atributos**: `nombre` y `raza` son las características de cada perro.
- **Método**: `ladrar()` es una acción que los objetos `Perro` pueden realizar.

## 5. Visualización de Datos con Matplotlib

Matplotlib es una de las librerías más populares para crear visualizaciones estáticas, animadas e interactivas en Python.

```python
# Es necesario tener instalada la librería: pip install matplotlib
import matplotlib.pyplot as plt
import numpy as np

# Datos de ejemplo
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Crear la gráfica
plt.figure(figsize=(8, 4))
plt.plot(x, y, label='sin(x)')
plt.title('Ejemplo simple con Matplotlib')
plt.xlabel('Eje X')
plt.ylabel('Eje Y')
plt.legend()
plt.grid(True)
plt.show()
```

## 6. Visualización de Grafos con Graphviz

Graphviz es una herramienta de software de visualización de grafos. Permite describir grafos en un lenguaje de texto simple y la herramienta genera diagramas en varios formatos útiles.

```python
# Es necesario tener instalada la librería: pip install graphviz
# Y también tener Graphviz instalado en el sistema: https://graphviz.org/download/
from graphviz import Digraph

# Crear un nuevo grafo dirigido para ilustrar la relación Clase -> Objeto
dot = Digraph(comment='Clase y Objetos')

# Añadir nodos
dot.node('A', 'Clase Perro')
dot.node('B', 'Objeto: mi_perro')
dot.node('C', 'Objeto: otro_perro')

# Añadir aristas (conexiones)
dot.edge('A', 'B', 'crea')
dot.edge('A', 'C', 'crea')

# Mostrar el grafo en el notebook (la última línea de una celda se muestra automáticamente)
dot
```
